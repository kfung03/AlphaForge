# Notebook 7 — Robustness Analysis

## Objective

The objective of this notebook is to evaluate whether the relative performance of Signature, GARCH, Combined, CNN, XGBoost, and Ensemble models remains stable across alternative experimental settings.

The robustness analysis will compare model performance across:

- Fama-French 3-factor and 5-factor inputs
- Monthly and daily frequencies
- Alternative rolling-window lengths
- Alternative signature depths
- Alternative top-portfolio classification thresholds
- Predictive and investment-performance metrics

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
MODEL_ROOT = Path("../models")
RESULT_ROOT = Path("../results")
ROBUSTNESS_DIR = RESULT_ROOT / "robustness"

ROBUSTNESS_DIR.mkdir(parents=True, exist_ok=True)
```

```python
EXPERIMENTS = {
    "monthly_ff3": {
        "frequency": "monthly",
        "factor_model": "FF3",
        "periods_per_year": 12
    },
    "monthly_ff5": {
        "frequency": "monthly",
        "factor_model": "FF5",
        "periods_per_year": 12
    },
    "daily_ff3": {
        "frequency": "daily",
        "factor_model": "FF3",
        "periods_per_year": 252
    },
    "daily_ff5": {
        "frequency": "daily",
        "factor_model": "FF5",
        "periods_per_year": 252
    }
}
```

```python
classification_results = []

for experiment_name, config in EXPERIMENTS.items():
    result_path = (
        MODEL_ROOT
        / experiment_name
        / "final_model_results.csv"
    )

    if not result_path.exists():
        print(f"Missing: {result_path}")
        continue

    df = pd.read_csv(result_path)

    df["experiment"] = experiment_name
    df["frequency"] = config["frequency"]
    df["factor_model"] = config["factor_model"]

    classification_results.append(df)

classification_results = pd.concat(
    classification_results,
    ignore_index=True
)

display(classification_results.head())
```
